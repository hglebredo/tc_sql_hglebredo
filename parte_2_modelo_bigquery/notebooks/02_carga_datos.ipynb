## Prerequisitos

Habiendo ejecutado previamente las instrucciones contenidas en [el setup de Big Query](01_setup_bigquery.ipynb), se procede a hacer la carga de datos.


```python
from google.cloud import bigquery
from google.oauth2 import service_account
import os
from dotenv import load_dotenv
load_dotenv()

# Ruta absoluta al proyecto
carpeta = os.path.abspath(os.path.join(os.getcwd(), os.pardir))

# Configuración
PROJECT_ID = os.getenv("GCP_PROJECT_ID")
DATASET_ID = os.getenv("BQ_DATASET_ID")
CREDENTIALS_PATH = os.path.join(carpeta, os.getenv("GOOGLE_APPLICATION_CREDENTIALS"))

# Cliente autenticado
credentials = service_account.Credentials.from_service_account_file(
    CREDENTIALS_PATH
)
client = bigquery.Client(
    project=PROJECT_ID,
    credentials=credentials
)
```
